```python
import pandas as pd
import numpy as np

# 1. Adatok betöltése a mappából
results_df = pd.read_csv('results.csv')
drivers_df = pd.read_csv('drivers.csv')
constructors_df = pd.read_csv('constructors.csv')

print("1. KÉRDÉS: Top 5 pilóta a legtöbb ponttal")
merged_drivers = pd.merge(results_df, drivers_df, on='driverId')
top_points = merged_drivers.groupby(['forename', 'surname'])['points'].sum().reset_index()
print(top_points.sort_values(by='points', ascending=False).head(5))

print("\n 2. KÉRDÉS: Átlagos célba érkezési pozíció a rajthely függvényében")
print(results_df.groupby('grid')['positionOrder'].mean().head(10))

print("\n 3. KÉRDÉS: Legtöbb győzelmet szerző konstruktőrök")
merged_const = pd.merge(results_df, constructors_df, on='constructorId')
wins = merged_const[merged_const['position'] == '1'].groupby('name')['position'].count().reset_index()
print(wins.sort_values(by='position', ascending=False).head(5))

print("\n 4. KÉRDÉS: Pozíciók átlaga és szórása")
positions = pd.to_numeric(results_df['position'], errors='coerce').to_numpy()
print(f"Átlagos célba érkezési pozíció: {np.nanmean(positions):.2f}")
print(f"Pozíciók szórása: {np.nanstd(positions):.2f}")

print("\n 5. KÉRDÉS: Korreláció a rajthely és a pontszám között")
correlation = results_df['grid'].corr(results_df['points'])
print(f"Rajthely és pontszám korrelációja: {correlation:.2f}")
```
