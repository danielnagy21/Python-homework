# Formula 1 Adatelemzési Szorgalmi Feladat

## 1. Bevezetés és Célkitűzés
A projekt célja egy Kaggle-ről származó Formula 1 adathalmaz feldolgozása és elemzése a Python programozási nyelv (`pandas` és `numpy` könyvtárak) segítségével. A vizsgálat során öt különböző kérdésre kerestem a választ a pilóták, konstruktőrök és a versenypozíciók teljesítménye alapján.

## 2. Felhasznált Adatok
Az elemzés a következő CSV fájlokat használta fel:
- `results.csv`: Az egyes versenyek eredményei és statisztikái.
- `drivers.csv`: A pilóták adatai (név, azonosító).
- `constructors.csv`: A konstruktőrök (csapatok) adatai.

## 3. Megfogalmazott Kérdések és Eredmények

### 1. Kérdés: Top 5 pilóta a legtöbb ponttal
- **Eredmény:** A történelem során a legtöbb pontot gyűjtő öt pilóta: Lewis Hamilton, Sebastian Vettel, Max Verstappen, Fernando Alonso és Kimi Räikkönen.

### 2. Kérdés: Átlagos célba érkezési pozíció a rajthely függvényében
- **Eredmény:** A számítások megmutatják, hogy az adott rajthelyről indulva átlagosan hányadik helyen érkeztek célba a pilóták.

### 3. Kérdés: Legtöbb győzelmet szerző konstruktőrök
- **Eredmény:** A legtöbb futamgyőzelmet elért csapatok listáját a Ferrari vezeti, a McLaren, a Mercedes, a Red Bull és a Williams előtt.

### 4. Kérdés: Pozíciók átlaga és szórása (Numpy használatával)
- **Eredmény:** Az elért célba érkezési pozíciók számtani közepe (átlaga) kb. 8.02, a szórása pedig 4.84 volt.

### 5. Kérdés: Korreláció a rajthely és a pontszám között
- **Eredmény:** A rajthely és a megszerzett pontszám közötti korreláció -0.38, ami enyhe negatív kapcsolatot mutat (az előkelőbb rajthely általában több pontot eredményez).
